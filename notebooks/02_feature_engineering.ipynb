```python
import sys
from pathlib import Path

import pandas as pd
import numpy as np

PROJECT_ROOT = Path.cwd().parent
sys.path.append(str(PROJECT_ROOT))

from src.data.data_loader import load_application_train

df = load_application_train()

print("Dataset shape:", df.shape)
print("Columns:", len(df.columns))
```

```python
df["LOAN_INCOME_RATIO"] = (
    df["AMT_CREDIT"] / df["AMT_INCOME_TOTAL"]
)

df["ANNUITY_INCOME_RATIO"] = (
    df["AMT_ANNUITY"] / df["AMT_INCOME_TOTAL"]
)

df["CREDIT_GOODS_RATIO"] = (
    df["AMT_CREDIT"] / df["AMT_GOODS_PRICE"]
)

print(
    df[
        [
            "LOAN_INCOME_RATIO",
            "ANNUITY_INCOME_RATIO",
            "CREDIT_GOODS_RATIO"
        ]
    ].describe()
)
```

```python
df["AGE_YEARS"] = (
    -df["DAYS_BIRTH"] / 365.25
)

df["DAYS_EMPLOYED_CLEAN"] = (
    df["DAYS_EMPLOYED"].replace(365243, np.nan)
)

df["EMPLOYMENT_YEARS"] = (
    -df["DAYS_EMPLOYED_CLEAN"] / 365.25
)

print(
    df[
        [
            "AGE_YEARS",
            "EMPLOYMENT_YEARS"
        ]
    ].describe()
)
```

```python
df["INCOME_PER_CHILD"] = (
    df["AMT_INCOME_TOTAL"] / (df["CNT_CHILDREN"] + 1)
)

df["INCOME_PER_FAMILY_MEMBER"] = (
    df["AMT_INCOME_TOTAL"] / (df["CNT_FAM_MEMBERS"] + 1)
)

print(
    df[
        [
            "INCOME_PER_CHILD",
            "INCOME_PER_FAMILY_MEMBER"
        ]
    ].describe()
)
```

```python
df["CREDIT_MINUS_INCOME"] = (
    df["AMT_CREDIT"] - df["AMT_INCOME_TOTAL"]
)

df["ANNUITY_MINUS_INCOME"] = (
    df["AMT_ANNUITY"] - df["AMT_INCOME_TOTAL"]
)

print(
    df[
        [
            "CREDIT_MINUS_INCOME",
            "ANNUITY_MINUS_INCOME"
        ]
    ].describe()
)
```

```python
df["EMPLOYMENT_AGE_RATIO"] = (
    df["EMPLOYMENT_YEARS"] / df["AGE_YEARS"]
)

print(
    df["EMPLOYMENT_AGE_RATIO"].describe()
)
```

```python
engineered_features = [
    "LOAN_INCOME_RATIO",
    "ANNUITY_INCOME_RATIO",
    "CREDIT_GOODS_RATIO",
    "AGE_YEARS",
    "DAYS_EMPLOYED_CLEAN",
    "EMPLOYMENT_YEARS",
    "INCOME_PER_CHILD",
    "INCOME_PER_FAMILY_MEMBER",
    "CREDIT_MINUS_INCOME",
    "ANNUITY_MINUS_INCOME",
    "EMPLOYMENT_AGE_RATIO",
]

print("Engineered features:", len(engineered_features))
print()

for feature in engineered_features:
    print(
        f"{feature}: "
        f"dtype={df[feature].dtype}, "
        f"missing={df[feature].isna().sum()}"
    )
```

```python
processed_path = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "application_train_features.csv"
)

df.to_csv(processed_path, index=False)

print("Saved:", processed_path)
print("Shape:", df.shape)
```

# Day 5 — Feature Engineering Summary

### Objective

The goal of Day 5 was to create meaningful financial, demographic, household, and employment-related features from the Home Credit application dataset.

### Engineered Features

1. **LOAN_INCOME_RATIO**
   - Measures the size of the requested credit relative to applicant income.
   - Higher values indicate a larger credit burden relative to income.

2. **ANNUITY_INCOME_RATIO**
   - Measures annual loan annuity relative to total income.
   - Helps capture repayment burden.

3. **CREDIT_GOODS_RATIO**
   - Compares requested credit amount with the price of the goods being financed.
   - Can help identify differences between financed amount and goods value.

4. **AGE_YEARS**
   - Converts `DAYS_BIRTH` into approximate age in years.
   - Provides a more interpretable representation of applicant age.

5. **DAYS_EMPLOYED_CLEAN**
   - Cleans the special `365243` value in `DAYS_EMPLOYED`.
   - The placeholder is replaced with `NaN` because it does not represent a valid employment duration.

6. **EMPLOYMENT_YEARS**
   - Converts cleaned employment duration from days into years.
   - Provides a more interpretable measure of employment experience.

7. **INCOME_PER_CHILD**
   - Calculates applicant income relative to the number of children.
   - Uses `CNT_CHILDREN + 1` to avoid division by zero.

8. **INCOME_PER_FAMILY_MEMBER**
   - Calculates income relative to the applicant's family size.
   - Uses `CNT_FAM_MEMBERS + 1` to avoid division by zero.

9. **CREDIT_MINUS_INCOME**
   - Difference between credit amount and total income.
   - Captures the absolute gap between borrowing amount and income.

10. **ANNUITY_MINUS_INCOME**
    - Difference between annuity amount and total income.
    - Provides another representation of payment burden.

11. **EMPLOYMENT_AGE_RATIO**
    - Measures employment duration relative to applicant age.
    - Provides an indication of how much of the applicant's life has been spent in employment.

### Feature Engineering Results

- Original columns: **122**
- Engineered features: **11**
- Final columns: **133**
- Training rows: **307,511**

### Missing Values

The employment-related missing values are expected because the original dataset contains a special placeholder value in `DAYS_EMPLOYED`. These missing values will be handled later during the model preprocessing stage.
